## Step 2: System Architecture & Resource Diagnostics
This step inspects Python runtime version, host operating system, allocated CPU cores, and total available RAM using standard Python modules (`sys`, `platform`, `os`, `psutil`).

```python
import sys
import platform
import os
import psutil

print("=" * 60)
print("🖥️  SYSTEM & ENVIRONMENT INFORMATION")
print("=" * 60)
print(f"Python Version    : {sys.version.split()[0]}")
print(f"Python Path       : {sys.executable}")
print(f"OS Platform       : {platform.platform()}")
print(f"Processor         : {platform.processor() or 'N/A'}")
print(f"CPU Cores (vCPU)  : {os.cpu_count()}")

ram_gb = psutil.virtual_memory().total / (1024 ** 3)
ram_avail_gb = psutil.virtual_memory().available / (1024 ** 3)
print(f"Total System RAM  : {ram_gb:.2f} GB")
print(f"Available RAM     : {ram_avail_gb:.2f} GB")
print("=" * 60)
```

## Step 4: Verify Package Versions & CUDA Compatibility
Now we perform an import test on every key package and display its installed version in a clean diagnostic table. We also verify that PyTorch is linked with CUDA support for accelerated training.

```python
import importlib

packages = [
    ("Python", sys.version.split()[0]),
    ("torch", None),
    ("torchvision", None),
    ("ultralytics", None),
    ("cv2 (OpenCV)", None),
    ("numpy", None),
    ("pandas", None),
    ("matplotlib", None),
    ("PIL (Pillow)", None),
    ("yaml (PyYAML)", None),
]

print("=" * 55)
print(f"{'PACKAGE NAME':<25} | {'INSTALLED VERSION':<25}")
print("=" * 55)

for pkg_name, default_ver in packages:
    if default_ver:
        print(f"{pkg_name:<25} | {default_ver:<25}")
        continue
    try:
        mod_name = pkg_name.split()[0]
        mod = importlib.import_module(mod_name)
        ver = getattr(mod, '__version__', 'Installed (Version N/A)')
        print(f"{pkg_name:<25} | {ver:<25}")
    except ImportError:
        print(f"{pkg_name:<25} | {'❌ NOT INSTALLED':<25}")

print("=" * 55)

# PyTorch CUDA Check
import torch

cuda_ok = torch.cuda.is_available()
print(f"\n🔥 CUDA Acceleration Available: {'✅ YES' if cuda_ok else '❌ NO (Running on CPU)'}")
if cuda_ok:
    print(f"   Device Count              : {torch.cuda.device_count()}")
    print(f"   Current Device Index      : {torch.cuda.current_device()}")
    print(f"   Device Name               : {torch.cuda.get_device_name(0)}")
    print(f"   CUDA Capability (SM)      : {torch.cuda.get_device_capability(0)}")
    vram_gb = torch.cuda.get_device_properties(0).total_memory / (1024 ** 3)
    print(f"   Total GPU Memory (VRAM)   : {vram_gb:.2f} GB")
else:
    print("   ⚠️  Please switch Colab runtime to GPU (Runtime > Change runtime type > T4 GPU).")
```
