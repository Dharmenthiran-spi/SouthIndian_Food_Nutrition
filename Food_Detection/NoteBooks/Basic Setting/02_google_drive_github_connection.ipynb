# 🔗 Connect Google Drive & Local PC via GitHub
**Project: South Indian Food Nutrition & Detection**

This notebook sets up a direct, two-way bridge between your **Local PC** (`D:\Yolo_Food_Inspection`) and your **Google Drive** folder using **GitHub** as the synchronization hub.

---

### 🔄 How The Sync Works:
```
Local PC (D:\Yolo_Food_Inspection)  <==== git push / pull ====>  GitHub Repo  <==== git push / pull ====>  Google Drive (/content/drive)
```

1. Any changes made on your **Local PC** are pushed to GitHub (`git push`).
2. Running **Step 3** in this notebook pulls those changes directly into your **Google Drive**.
3. Any changes made in **Google Drive / Colab** can be committed and pushed to GitHub (**Step 4**) and pulled to your PC (`git pull`).
