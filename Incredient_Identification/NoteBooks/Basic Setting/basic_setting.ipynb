## 2. Install & Verify Package Versions
Install all necessary deep learning and computer vision libraries, and print an audit table of exact installed versions.

````python
# Install essential project packages
!pip install -q ultralytics torchvision opencv-python-headless pyyaml matplotlib seaborn pandas pillow psutil
````

````python
import importlib, torch

packages = [
    ("Python", sys.version.split()[0]),
    ("torch", None),
    ("torchvision", None),
    ("ultralytics", None),
    ("cv2 (OpenCV)", None),
    ("numpy", None),
    ("pandas", None),
    ("matplotlib", None),
    ("PIL (Pillow)", None),
    ("yaml (PyYAML)", None),
]

print("=" * 55)
print(f"{'PACKAGE NAME':<25} | {'INSTALLED VERSION':<25}")
print("=" * 55)
for pkg_name, default_ver in packages:
    if default_ver:
        print(f"{pkg_name:<25} | {default_ver:<25}")
        continue
    try:
        mod = importlib.import_module(pkg_name.split()[0])
        ver = getattr(mod, '__version__', 'Installed')
        print(f"{pkg_name:<25} | {ver:<25}")
    except ImportError:
        print(f"{pkg_name:<25} | {'❌ NOT INSTALLED':<25}")
print("=" * 55)

# CUDA Check
cuda_ok = torch.cuda.is_available()
print(f"\n🔥 CUDA Acceleration : {'✅ ACTIVE' if cuda_ok else '❌ INACTIVE (Switch Colab runtime to GPU)'}")
if cuda_ok:
    print(f"   GPU Device Name   : {torch.cuda.get_device_name(0)}")
    vram = torch.cuda.get_device_properties(0).total_memory / (1024 ** 3)
    print(f"   Total VRAM        : {vram:.2f} GB")
````

### 🟢 PART A: Running in Google Colab (Google Drive)

#### 📥 A1. GIT PULL IN GOOGLE DRIVE (Receive updates from Local PC)
Run this cell whenever you push changes from your local computer and want them in Google Drive:

````python
# Bulletproof PULL from GitHub into Google Drive
# Resets any Colab autosave metadata and pulls latest code
!git fetch origin main
!git reset --hard origin/main
print('✅ Google Drive successfully synchronized with GitHub!')
````

#### 📤 A2. GIT PUSH FROM GOOGLE DRIVE (Send updates to Local PC via GitHub)
Run this cell whenever you create or edit notebooks/files in Google Drive and want to send them to GitHub and your Local PC:

````python
# PUSH from Google Drive to GitHub
!git add .
!git commit -m "Update from Google Drive"
!git push origin main
````
